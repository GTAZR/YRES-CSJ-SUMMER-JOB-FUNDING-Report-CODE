````python
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils import get_column_letter

output_path = r'C:\Users\admin\Desktop\YRES\Andi Dong - Summer Job Analysis\CSJ_Analysis_Output.xlsx'

# 定义6个sheet的配置
sheets = {
    'ON ridings':       on_ridings,
    'ON organizations': on_orgs,
    'BC ridings':       bc_ridings,
    'BC organizations': bc_orgs,
    'AB ridings':       ab_ridings,
    'AB organizations': ab_orgs
}

# Step 1: 用pandas先导出基础数据
with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    for sheet_name, df in sheets.items():
        df.to_excel(writer, sheet_name=sheet_name, index=False)

# Step 2: 用openpyxl加格式
wb = load_workbook(output_path)

for sheet_name in wb.sheetnames:
    ws = wb[sheet_name]
    
    # 表头加粗 + 背景色 + 字体颜色
    header_font = Font(bold=True, color='FFFFFF', name='Arial')
    header_fill = PatternFill('solid', start_color='2F5597')  # 深蓝色
    
    for cell in ws[1]:
        cell.font = header_font
        cell.fill = header_fill
        cell.alignment = Alignment(horizontal='center', wrap_text=True)
    
    # 冻结首行
    ws.freeze_panes = 'A2'
    
    # 列宽自适应
    for col in ws.columns:
        max_length = 0
        col_letter = get_column_letter(col[0].column)
        for cell in col:
            if cell.value:
                max_length = max(max_length, len(str(cell.value)))
        ws.column_dimensions[col_letter].width = min(max_length + 2, 30)
    
    # 货币格式：所有funding列
    for row in ws.iter_rows(min_row=2):
        for cell in row:
            col_name = ws.cell(1, cell.column).value
            if col_name and 'funding' in str(col_name):
                cell.number_format = '$#,##0'
            elif col_name and 'avg_salary' in str(col_name):
                cell.number_format = '$#,##0.00'



wb.save(output_path)
print("专业Excel导出完成！")

'''
**代码做了什么：**
- 深蓝色表头 + 白色加粗字体
- 冻结首行，滚动时表头不动
- 列宽自适应，最宽30
- funding列加`$`货币格式
- avg_salary列加`$`保留两位小数
'''
````

### 最后来一步骤，简单做一个线性回归预测模型！目的在于预测所有的选区，记得不是机构！总共190多个选区下，根据17-25年所有数据，来分析每个选区的未来funding走势来判断后续可以理论性重点考虑投资哪一个选区最为合适？

````python
stg_csj
````

````python
# 按照省份，选区，年来分组！
constituency_yearly = stg_csj.groupby(['region','constituency','year']).agg(total_funding = ('amount','sum'), total_jobs = ('jobs', 'sum')).reset_index()
````

````python
constituency_yearly
````

````python
constituency_yearly.head(50)
````

````python
from sklearn.linear_model import LinearRegression
import numpy as np

results = []

#每个选区group单独训练模型！
for (region, constituency), group in constituency_yearly.groupby(['region', 'constituency']):
    X = group['year'].values.reshape(-1,1) # year as features!
    y_funding = group['total_funding'].values # funding as the target 1
    y_jobs = group['total_jobs'].values # jobs created as the target 2

    # train the model
    model_funding = LinearRegression().fit(X, y_funding)
    model_jobs = LinearRegression().fit(X, y_jobs)

    # predict for values in 2026
    pred_funding = model_funding.predict([[2026]])[0]
    pred_jobs = model_jobs.predict([[2026]])[0]

    results.append({
    'region': region,
    'constituency': constituency,
    'pred_funding_2026': round(max(pred_funding, 0)),
    'pred_jobs_2026': round(max(pred_jobs, 0))
    })

predictions = pd.DataFrame(results)
print(predictions.head(10))
````

````python
with pd.ExcelWriter(output_path, mode='a', engine='openpyxl') as writer:
    predictions.to_excel(writer, sheet_name='2026 Predictions', index=False)

print("预测sheet添加完成！")
````

````python
# 把predictions整理成和all_ridings_long一样的行格式
pred_rows = predictions.copy()
pred_rows = pred_rows.rename(columns={
    'region': 'province',
    'pred_funding_2026': 'funding',
    'pred_jobs_2026': 'jobs'
})
pred_rows['year'] = 2026
pred_rows['avg_salary'] = (pred_rows['funding'] / pred_rows['jobs']).round(2)
pred_rows['total_number_of_organizations'] = None
pred_rows['is_predicted'] = True

# 原始数据加标记
all_ridings_long['is_predicted'] = False

# 拼接
all_ridings_long = pd.concat(
    [all_ridings_long, pred_rows[all_ridings_long.columns]], 
    ignore_index=True
)

print(all_ridings_long.shape)  # 应该是1980行
````

# Python Layer - CSJ Analysis Pipeline
**目标：SQL做不到的自动化处理，为Tableau和YRES交付做准备**

---

## Step 1: 连接DuckDB，读取6张mart表
- 连接 `csj_analysis.duckdb`
- 读取 `mart_on/bc/ab_ridings` 和 `mart_on/bc/ab_organizations`
- 验证数据形状：ON(121,29) BC(42,29) AB(35,29)

---

## Step 2: 数据清洗
- 找出所有 `avg_salary` 列
- 统一四舍五入到2位小数
- 用for loop批量处理6张表

---

## Step 3: 加province列 + concat拼接
- 6张表各自加 `province` 列标记省份
- riding三表合并 → `all_ridings` (198行 × 30列)
- org三表合并   → `all_orgs`   (5911行 × 30列)

---

## Step 4: 宽表转长表（pd.melt + pivot_table）
**为什么要转：Tableau需要year作为独立列才能画趋势图**

- `pd.melt()` 把年份列转成行
- `str.rsplit('_', n=1)` 拆分metric和year
- `pivot_table()` 把funding/jobs/avg_salary变回三列
- 结果：
  - `all_ridings_long` (1782行 × 7列)
  - `all_orgs_long`    (53199行 × 7列)

---

## Step 5: 线性回归预测2026
**为什么用线性回归：每个选区只有9年数据，复杂模型会过拟合**

- 从 `stg_csj` 用pandas groupby得到选区年度汇总
- 对198个选区各自训练独立的LinearRegression模型
- X = year, y = total_funding / total_jobs
- 预测2026年每个选区的funding和jobs
- `max(pred, 0)` 确保预测值不为负

---

## Step 6: 预测结果整合
- 预测结果加入Excel作为第7个sheet（`2026 Predictions`）
- 预测结果转成长表格式，加 `is_predicted=True` 标记
- 拼接进 `all_ridings_long`
- 最终：1980行（1782历史 + 198预测）

---

## Step 7: 导出Excel（YRES交付）
- 6个sheet对应6张分析表
- 格式：深蓝表头 + 白色加粗字体 + 冻结首行 + 列宽自适应
- funding列：`$#,##0` 货币格式
- avg_salary列：`$#,##0.00` 格式
- 第7个sheet：2026预测结果

---

## Step 8: 导出Tableau CSV（3个文件）
| 文件名 | 内容 | 用途 |
|--------|------|------|
| `tableau_ridings_long.csv` | 198选区 × 10年(含2026预测) | 趋势图/预测对比 |
| `tableau_orgs_top30_long.csv` | TOP30组织历史数据 | 竞争对手分析 |
| `tableau_all_orgs_full.csv` | 全量原始数据 | 深度分析 |
